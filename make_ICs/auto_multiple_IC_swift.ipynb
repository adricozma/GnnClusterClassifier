# Auto SWIFT IC maker from an array of initial boost positions and velocities

```python
import numpy as np
import pandas as pd
```

```python
#importing the npy data
o = np.load('SWIFT_seeds.npy', allow_pickle=True)
# creating a dataframe with the initial conditions
df_seeds = pd.DataFrame({
    'x_kpc': o[:,1],
    'y_kpc': o[:,2],
    'z_kpc': o[:,3],
    'vx_kms': o[:,4],
    'vy_kms': o[:,5],
    'vz_kms': o[:,6],
    'Stream_ID': o[:,0]
})
print("done...")
print(df_seeds.head())  
```

```python
# launching the file that generates the initial conditions for SWIFT
import os
from IC_plummer_swift import generate_plummer
n=len(df_seeds)
N = np.random.randint(20000, 50000, n)
M = N * 1.0 # mass of each particle in solar masses
boxsize = 1000 # size of the box in kpc
folder = "initial_conditions/"


#looping over all the seeds and generating the initial conditions for each stream
i = 0
for row in df_seeds.itertuples():
    print(i)
    a_value = np.random.uniform(0.001, 0.05)
    sid = row.Stream_ID
    x_prog = row.x_kpc
    y_prog = row.y_kpc
    z_prog = row.z_kpc
    vx_prog = row.vx_kms
    vy_prog = row.vy_kms
    vz_prog = row.vz_kms
    name = folder + f"IC_{int(sid)}.hdf5"
    if not os.path.exists(name):
        _ = generate_plummer(a=a_value, M=M[i], q=1.0, N=N[i], bound=2.0, boxsize=boxsize, 
                                use_parallel=False, nb_threads=2, 
                                shiftx=x_prog, shifty=y_prog, shiftz=z_prog, 
                                shiftvx=vx_prog, shiftvy=vy_prog, shiftvz=vz_prog,
                                fname=name)
    i +=1
```
